```python
for name, df in {
    "Source 1": train_s1,
    "Source 2": train_s2,
    "Source 3": train_s3
}.items():

    print(f"\n===== {name} =====")

    for col in ["business_name", "business_address", "country"]:
        empty_count = (
            df[col]
            .fillna("")
            .astype(str)
            .str.strip()
            .eq("")
            .sum()
        )

        print(f"{col}: {empty_count} empty values")
```

```python
sample_gt = train_s1[["entity_id"]].merge(
    ground_truth,
    left_on="entity_id",
    right_on="source1_entity_id",
    how="left"
)

display(sample_gt.head(20))
```

```python
def count_matches(value):
    if pd.isna(value) or str(value).strip() == "":
        return 0

    return len(
        [x for x in str(value).split(",") if x.strip()]
    )

sample_gt["match_count"] = sample_gt["matched_entity_ids"].apply(count_matches)

display(
    sample_gt[["source1_entity_id", "matched_entity_ids", "match_count"]].head(20)
)
```

```python
def match_sources(value):
    if pd.isna(value) or str(value).strip() == "":
        return "none"

    ids = [x.strip() for x in str(value).split(",") if x.strip()]

    has_s2 = any(x.startswith("S2-") for x in ids)
    has_s3 = any(x.startswith("S3-") for x in ids)

    if has_s2 and has_s3:
        return "S2+S3"
    elif has_s2:
        return "S2"
    elif has_s3:
        return "S3"
    else:
        return "other"

sample_gt["match_source_type"] = sample_gt["matched_entity_ids"].apply(match_sources)

print(sample_gt["match_source_type"].value_counts())
```
