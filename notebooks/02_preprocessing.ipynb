```python
def normalize_text(value):
    """
    Basic Unicode-safe text normalization.

    Steps:
    1. Handle missing values
    2. Unicode normalization
    3. Case normalization
    4. Replace punctuation/symbols with spaces
    5. Collapse repeated whitespace
    """
    if pd.isna(value):
        return ""

    text = str(value)

    # Normalize Unicode representation
    text = unicodedata.normalize("NFKC", text)

    # Case-insensitive normalization while preserving Unicode
    text = text.casefold()

    # Replace punctuation/symbols with spaces.
    # \w preserves Unicode word characters.
    text = re.sub(r"[^\w\s]", " ", text, flags=re.UNICODE)

    # Collapse repeated whitespace
    text = re.sub(r"\s+", " ", text).strip()

    return text
```

```python
test_values = [
    "ABC Pvt. Ltd.",
    "ABC   PHARMACY",
    "A.B.C Pharmacy",
    "Shivshakti Vidyalaya",
    "International South Consultants Private Ltd",
    None
]

for value in test_values:
    print(f"{value!r}  ->  {normalize_text(value)!r}")
```

```python
def normalize_name(value):
    """
    Normalize business names.
    """
    return normalize_text(value)


def normalize_address(value):
    """
    Normalize business addresses.
    """
    return normalize_text(value)


def normalize_country(value):
    """
    Normalize country values.
    """
    if pd.isna(value):
        return ""

    text = unicodedata.normalize("NFKC", str(value))
    text = text.casefold()
    text = re.sub(r"\s+", " ", text).strip()

    return text
```

```python
s1_clean = train_s1.copy()
s2_clean = train_s2.copy()
s3_clean = train_s3.copy()
```

```python
for df in [s1_clean, s2_clean, s3_clean]:
    df["business_name_clean"] = df["business_name"].apply(normalize_name)
    df["business_address_clean"] = df["business_address"].apply(normalize_address)
    df["country_clean"] = df["country"].apply(normalize_country)
```
